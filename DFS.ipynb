```python
# DFS (Depth First Search)

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

visited = set()

def DFS(node):
    if node not in visited:
        print(node, end=" ")
        visited.add(node)

        for neighbour in graph[node]:
            DFS(neighbour)

print("DFS Traversal:")
DFS('A')
```

DLS Traversal: A → B → D → E → C → F → G
Depth Limit: 2
