```python


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

def DLS(node, limit, depth=0):
    print(node, end=" ")

    if depth == limit:
        return

    for neighbour in graph[node]:
        DLS(neighbour, limit, depth + 1)

print("DLS Traversal:")
DLS('A', 2)
```
