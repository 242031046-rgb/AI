```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

def dls(start, limit):
    stack = [(start, 0)]

    while stack:
        node, depth = stack.pop()
        print(node, end=" ")

        if depth < limit:
            for child in reversed(graph[node]):
                stack.append((child, depth + 1))

print("DLS Traversal:")
dls('A', 2)
```

Depth Limit: 2
