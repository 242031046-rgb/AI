```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

def dls(node, limit, depth=0):
    print(node, end=" ")

    if depth == limit:
        return

    for child in graph[node]:
        dls(child, limit, depth + 1)

print("IDS Traversal:")

for depth in range(3):
    print("\nDepth Limit:", depth)
    dls('A', depth)
```

Depth 0: A
