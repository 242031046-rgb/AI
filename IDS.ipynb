```python
def dls(node, target, limit, graph, visited_path):
    """Depth-Limited Search (DLS) helper function"""
    if node == target:
        return [node]

    if limit <= 0:
        return None

    for neighbor in graph.get(node, []):
        if neighbor not in visited_path:
            result = dls(neighbor, target, limit - 1, graph, visited_path + [node])
            if result is not None:
                return [node] + result
    return None

def ids(graph, start, target, max_depth):
    """Iterative Deepening Search (IDS) main function"""
    for depth in range(max_depth + 1):
        print(f"--- Searching at Depth Limit: {depth} ---")
        path = dls(start, target, depth, graph, [])
        if path is not None:
            print(f"-> Success! Target found at depth {depth}.")
            return path
        else:
            print(f"-> Target not found within depth {depth}, increasing limit...")
    return None

# Example Graph (Adjacency List)
graph = {
    'S': ['A', 'B'],
    'A': ['C', 'D'],
    'B': ['E', 'F'],
    'C': [],
    'D': [],
    'E': [],
    'F': []
}

# Run IDS from 'S' to target 'F'
start_node = 'S'
target_node = 'F'
max_allowed_depth = 3

solution_path = ids(graph, start_node, target_node, max_allowed_depth)

print("\n==============================")
print("Final Result Path:", solution_path)
print("==============================")
```
