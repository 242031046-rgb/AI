```python
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 2), ('G', 5)],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

visited = []
queue = [('A', 0)]

while queue:
    queue.sort(key=lambda x: x[1])
    node, cost = queue.pop(0)

    if node not in visited:
        print(node, "Cost:", cost)
        visited.append(node)

        for child, weight in graph[node]:
            if child not in visited:
                queue.append((child, cost + weight))
```
